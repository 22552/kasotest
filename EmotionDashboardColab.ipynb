# 第二プロジェクト 感情・活動ダッシュボード

新DB・半年前DB・added・removed を対象に、ユーザー別の活動時期と8感情の文章表現を分析します。

- **活動分析は対象ユーザーの全コメントを使用**: 日別コメント数 / 0〜23時の棒グラフ / 曜日×時間帯ヒートマップ
- **感情分析だけサンプリング可能**: 喜び・悲しみ・期待・驚き・怒り・恐れ・嫌悪・信頼
- 活動分析は先に計算して即表示し、その後に感情分析結果を追記します
- 感情ごとのユーザーランキング / 感情スコアでコメントをフィルター

> 感情スコアは文章表現のモデル推定値で、本人の心理状態や性格を示すものではありません。
